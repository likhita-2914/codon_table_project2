```python
dna_sequence = input("Enter a DNA sequence: ")
dna_sequence = dna_sequence.strip().upper()

print("Input sequence:", dna_sequence)
```

function for accepting the DNA seq from the user

```python
def validate_dna(sequence):
    valid_bases = set("ATGC")

    if not sequence:
        return False

    return all(base in valid_bases for base in sequence)

if validate_dna(dna_sequence):
    print("Valid DNA sequence")
else:
    print("Invalid DNA sequence. Use only A, T, G and C.")
```

qc of the input seq

```python
def split_into_codons(sequence):
    codons = []

    for i in range(0, len(sequence), 3):
        codon = sequence[i:i + 3]
        codons.append(codon)

    return codons

codons = split_into_codons(dna_sequence)

print(codons)
```

split seq to codons

```python
def translate_codons(codons, table):
    amino_acids = []

    for codon in codons:
        if len(codon) != 3:
            amino_acids.append("Incomplete codon")
        else:
            amino_acids.append(table[codon])

    return amino_acids

amino_acids = translate_codons(codons,codon_table)

print(amino_acids)
```

translate the seq


```python
START_CODON = "ATG"
STOP_CODONS = {"TAA", "TAG", "TGA"}


def identify_start_stop(codons):
    start_codons = []
    stop_codons = []

    for index, codon in enumerate (codons):
        if codon == START_CODON:
            start_codons.append((index + 1, codon))

        if codon in STOP_CODONS:
            stop_codons.append((index + 1, codon))

    return start_codons, stop_codons
```

```python
codons = codons

starts, stops = identify_start_stop(codons)

print("Start codons:", starts)
print("Stop codons:", stops)
```

identify and count start and stop codons

```python
def separate_incomplete_codons(codons):
    complete = [codon for codon in codons if len(codon) == 3]
    incomplete = [codon for codon in codons if len(codon) != 3]

    return complete, incomplete

codons = split_into_codons(dna_sequence)
complete, incomplete = separate_incomplete_codons(codons)

print("Complete codons:", complete)
print("Incomplete codons:", incomplete)
```

handle incomplete codons

```python
def display_results(codons, amino_acids):
    print("Position | Codon | Amino Acid")
    print("-----------------------------")

    for i in range(len(codons)):
        print(i + 1, "|", codons[i], "|", amino_acids[i])
display_results(codons, amino_acids)
print("Amino acid sequence:", " - ".join(amino_acids))
```

```python
rna_sequence = input("Enter an RNA sequence: ")
rna_sequence = rna_sequence.strip().upper()

print("Input sequence:", rna_sequence)
```

```python
def validate_rna(sequence):
    valid_bases = set("AUGC")

    if not sequence:
        return False

    return all(base in valid_bases for base in sequence)

if validate_rna(rna_sequence):
    print("Valid RNA sequence")
else:
    print("Invalid RNA sequence. Use only A, U, G and C.")
```

```python
def translate_rna(seq):
    sequence = rna_sequence

    if not sequence or any(base not in "AUGC" for base in sequence):
        raise ValueError("RNA must contain only A, U, G and C.")

    dna_equivalent = sequence.replace("U", "T")
    codons = split_into_codons(dna_equivalent)
    amino_acids = translate_codons(codons)

    return codons, amino_acids


print(codons)
print(amino_acids)
```
